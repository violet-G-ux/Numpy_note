```python
##直方图
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
pen=pd.read_csv('data/penguins.csv')
#data 要传DataFrame 对象
'''
- `data`：DataFrame（表格对象，**不是文件路径字符串**）
- `x`：x 轴的列名（字符串），做纵向直方图
- `y`：y 轴的列名（字符串），做横向直方图
- `bins`：柱子数量，整数
- `kde=False`：是否叠加密度曲线，`kde=True` 打开
- `hue`：分组列名，自动按颜色分组
- `alpha`：透明度，0~1
'''
sns.histplot(data=pen,x="bill_length_mm",kde=True)
```

```python
##计数图
sns.countplot(data=pen,x="island")
```

```python
##散点图
sns.scatterplot(data=pen,x="body_mass_g",y="flipper_length_mm",hue="sex")
```

```python
##蜂窝图
sns.jointplot(data=pen,x="body_mass_g",y="flipper_length_mm",kind="hex")
```

```python
##成对关系图
#把所有数值变量两两配对，一次性画出全部组合关系
sns.pairplot(data=pen,hue="species")
```
