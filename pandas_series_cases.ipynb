```python
'''
创建一个包含 10 名学生数学成绩的 Series，成绩范围在 50-100 之间。
计算平均分、最高分、最低分，并找出高于平均分的学生人数。
'''
import pandas as pd
import numpy as np
arr=np.random.randint(50,101,10)#左闭右开区间
s=pd.Series(arr,index=['学生' +str(i) for i in range(1,11)])#str(i)把数字转换成字符
print(s)
s1=s.mean()
print(f"平均分：{s1},最高分：{s.max()},最低分：{s.min()},高于平均分人数：{(s>s1).sum()}")
#人数这个其他做法：
print(s[s>s1].count())
print(len(s[s>s1]))#series和array相比多了索引，array的很多方法都还是适用的
```

```python
'''
给定某城市一周每天的最高温度Series，完成以下任务：
找出温度超过30度的天数
计算平均温度
将温度从高到低排序
找出温度变化最大的两天
temperatures = pd.Series([28, 31, 29, 32, 30, 27, 33],
index=['周一','周二','周三','周四','周五','周六','周日'])
'''
t = pd.Series([28, 31, 29, 32, 30, 27, 33],index=['周一','周二','周三','周四','周五','周六','周日'])
print(f"超过30度的天数:{t[t>30].count()},\n平均温度：{t.mean()}")
t1=t.sort_values(ascending=False)#升序为false所以是降序
print(f"排序后温度：{t1}")

##计算series数组的变化值
t2=t.diff().abs()#先调用diff方法再对得到的数列取绝对值
t3=t2.sort_values(ascending=False)
print(*(t3.head(2).keys()))#前面加星号可以把列表的中括号去掉
```

```python
'''
给定某股票连续 10 个交易日的收盘价 Series：
计算每日收益率（当日收盘价 / 前日收盘价 - 1）
找出收益率最高和最低的日期
计算波动率（收益率的标准差）
prices = pd.Series([102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1, 109.3, 110.2], index=pd.date_range('2023-01-01', periods=10))
'''
prices = pd.Series([102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1, 109.3, 110.2], index=pd.date_range('2023-01-01', periods=10))##这个索引的意思是生成一个从2023年1月1日开始的10个连续日期

shouyi=prices.pct_change()
print(shouyi.idxmax())
print(shouyi.idxmin())
print(shouyi.std())
```

```python
'''某产品过去 12 个月的销售量 Series：
计算季度平均销量（每 3 个月为一个季度）
找出销量最高的月份
计算月环比增长率
找出连续增长超过 2 个月的月份

sales = pd.Series([120, 135, 145, 160, 155, 170, 180, 175, 190, 200, 210, 220],index=pd.date_range('2022-01-01', periods=12, freq='M'))
'''
sales = pd.Series([120, 135, 145, 160, 155, 170, 180, 175, 190, 200, 210, 220],index=pd.date_range('2022-01-01', periods=12, freq='M'))
'''
### 常用 freq 简写
- `'D'`：Day，按天（默认）
-'M' 按月
- `'ME'`：月末
- `'MS'`：MonthStart，月初
- `'Q'`：Quarter，季度末
- `'H'`：Hour，小时'''

###resample('Q')`：Q = 季度，把每 3 个月分成一组,只是分组了，对每个分组要进行的操作也要明确
sales1=sales.resample('Q').mean()
print(sales1)
print(sales.idxmax())
##pct_change()月环比增长率：和紧挨着的上一个月比，增长了百分之多少，(当前值 − 上一期) / 上一期
##diff差值：比上个月多了多少，当前值 − 上一期的值
sales2=sales.pct_change()
print(sales2)

##找出连续增长超过 2 个月的月份
sales3=sales2>0#得到一个布尔数组
print(sales3)
#用一个大小为3的窗口去滑动，如果窗口里都为true就满足
print('-'*10)
sales3[sales3.rolling(3).sum()==3].keys().tolist()
```

```python
'''某商店每小时销售额 Series：
按天重采样计算每日总销售额
计算每天营业时间（8:00-22:00）和非营业时间的销售额比例
找出销售额最高的 3 个小时

np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0, 100, 24),
index=pd.date_range('2025-01-01', periods=24, freq='H'))
'''
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0, 100, 24),
index=pd.date_range('2025-01-01', periods=24, freq='H'))
print(hourly_sales)

##按天重采样计算每日总销售额
hour1=hourly_sales.resample('D').sum()
print(hour1)

##计算每天营业时间（8:00-22:00）

#第一种方法：
#hourly_sales.between_time('8:00','22:00')
#第二种
print('-'*20)
print(hourly_sales[(hourly_sales.index.hour>=8)&(hourly_sales.index.hour<=22)])
##非营业时间的销售额
busi=hourly_sales.between_time('8:00','22:00')
busis=busi.sum()
not_busi=hourly_sales.drop(busi.index).sum()
#或 not_busi=hourly_sales[(...)|(...)]
#非号是 ~ 有如下：(...)&(...)==~(...)|(...)
##比例
print(busis/not_busi)
print('-'*100)
##找出销售额最高的 3 个小时
hour4=hourly_sales.sort_values(ascending=False).head(3)
##或者 用.largest(3)
print(hour4)
```
