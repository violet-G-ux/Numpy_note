```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.sans-serif'] = ['SimHei']

##导入数据
df=pd.read_csv('data/weather.csv')
# print(df.head())

##绘制气温的趋势变化图
df['date'] = pd.to_datetime(df['date'])
# df.info()
df=df[df['date'].dt.year==2023]


df_month_start = df[df['date'].dt.day == 1]

plt.figure(figsize=(10,8))
# 现在x、y都用月初筛选后的数据集
plt.plot(df_month_start['date'],df_month_start['max_temp'],label='max_temp')
plt.plot(df_month_start['date'],df_month_start['min_temp'],label='min_temp')
# 补上图例、标题、坐标轴
plt.legend(loc='upper left')
plt.title('2023年气温变化趋势')
plt.xlabel('日期')
plt.ylabel('气温')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
