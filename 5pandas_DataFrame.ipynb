## 一 .DataFrame的创建

```python
#这个DataFrame是在Series的基础上的，多行多列的带有行标签和列标签的二维的数组
#DataFrame的创建
import pandas as pd
import numpy as np

##通过Series来创建
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([6,7,8,9,10])
df=pd.DataFrame({"第一列":s1,"第二列":s2})  ##注意里面是字典
print(df)

##通过字典来创建
df=pd.DataFrame({
    "id":['001','001','003','004','005','006'],
    "name":["tom","tom","jac","alic",'bob','alien'],
    "age":[22,22,25,33,42,73],
    "score":[67,67,4,5,5,88]
},index=[1,2,3,4,5,6],columns=["name","age","id","score"])##如果想修改列的顺序可以直接在columns里面进行修改，就不需要改变原来字典里面的顺序
print(df)

##通过列表创建
df1=pd.DataFrame([[1,2,3,34],[2,3,4,5]])
df1
```

## 二. DataFrame的属性

```python
#DataFrame的属性
print('行索引：',df.index)
print('列标签:',df.columns)
print('DataFrame的值：',df.values)
print('维度：',df.ndim)
print('数据类型：',df.dtypes)
print('形状：',df.shape)
print('数据个数：',df.size)
print('行列转置：',df.T)
```

```python
##获取元素 loc iloc at iat

##显式获取
#某行
print(df.loc[4])#获取第四行数据
#某列
print(df.loc[:,'name'])#不能只写一个name
#某个元素
print(df.at[3,'name'])
print("隐式获取：")

##隐式获取
#获取行
print(df.iloc[3]) ##隐式获取时，索引是从零开始的
#获取列
print(df.iloc[:,1])
#获取某个元素
print(df.iat[2,1])
```

## 三.获取DataFrame的数据

```python
##获取单列数据
print(df['name']) ##这样得到的其实是 Series类型
print(df.age) ##不用中括号，直接点 获取出来是一样的效果 也是Series类型
print(df[['name']]) ##这样得到的不是Series而是DataFrame类型
```

```python
##查看多行数据
print(df.head(2))
print(df.tail(2))

##查看多列数据
print(df[['name','age']]) ##不能只写一个中括号，一个中括号只能用来获取一列数据
```

```python
##使用布尔索引
df1=df[df.age>10]
df2=df[(df.age<75)&(df.score>30)]
print(df2)
```

```python
##随机取样
df.sample(3) ##括号写几就是取几条，不写就是一条
```

## 四.DataFrame的常用方法

```python
print(df)
#查看元素是否包含在参数列表中
print(df.isin(['jac',25]))
```

```python
#查看元素是否是缺失值
print(df.isna())
```

```python
#sum
print(df['score'].sum())#一般就是求某一列的
print(df.score.max()) ##这个对于指定某一列是有两种方法的
print(df.age.min())
print(df.score.mean())
print(df.score.median())
print("众数：",df.score.mode())##众数
print(df.score.std())
print(df.score.var())
print(df.score.quantile(0.75))##分位数
##打印全部的基础信息
print(df.describe())
```

```python
print(df.value_counts())##这个是统计每行出现的次数，如果有行与行之间是相同的那么就会计数，如果只是有某个是相同的并不会认为是一样的
```

```python
##去重
print(df.drop_duplicates())
##检查是否是重复的
print(df.duplicated())
##检查某列是否有重复值
print(df.duplicated(subset=['age']))
```

```python
##把整张dataframe中的某个数值全部替换成另一个
print(df.replace(22,30))
```

```python
##累积和
df.cumsum()
df['age'].cummax()##这个cumsum是走一步更新一次最大值
##排序
print(df.sort_values(by='age',ascending=False))##这个按值排序的时候要指明按哪列
```

```python
##两列按照不同的要求排序
print(df.sort_values(by=['score','age'],ascending=[False,True]))
```

```python
##求最大值
df.nlargest(2,columns=['score','age'])##前面那个参数是取出来多少列，后面是按照那两列取

##求最小值
df.nsmallest(2,columns=['score','age'])
```
