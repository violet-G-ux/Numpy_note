### 一 Series的创建

```python
###series就是在ndarray上添加了索引
##Series = ndarray 数据 + 自定义标签索引 index
#series的创建
##通过列表创建Series
import pandas as pd
s=pd.Series([11,28,3,4,5])
print(s)
'''
索引 值
0    11
1    28
2     3
3     4
4     5
'''
#自定义索引
s=pd.Series([10,2,5,6,9 ],index=['a','b','c','d','e'])
print(s)

#定义name（用来描述'值'是干什么的）
s1=pd.Series([10,2,3,4,5],index=['a','b','c','d','e'],name='月份')
print(s1)
print('-'*10)
##通过字典创建Series
s=pd.Series({"a":12,"b":23,"c":34,"d":34,"e":35,"f":92,"g":88})
print(s)

print('-'*10)
#在已有的Series基础上创建Series(根据索引)
s2=pd.Series(s,index=["a","e"])
print(s2)
```

### 二 Series的属性

```python
'''
index:Series的索引对象
values:Series的值
dtype或dtypes:"Series的元素类型
shape:Series的形状
ndim:Series的维度
size:Series的元素个数
name:Series的名称

loc[]  显式索引，按标签索引或切片
iloc[] 隐式索引，按位置索引或切片
at[]   使用标签访问单个元素
iat[]  使用位置访问单个元素
'''
print(s,'\n')
print(s.index,'\n')
print(s.values,'\n')
print("形状：",s.shape,'\n')
print("维度：",s.ndim,"大小：",s.size)
print("Series中的数据类型：",s.dtypes)

s.name="lalala"
print("Series的名字：",s.name)

#显式索引:索引是啥就是啥
print(s.loc['b'])
#隐式索引：0-1-2-3-4-5---
print(s.iloc[2])

#索引切片
print(s.loc['a':'c'])#左闭右闭区间

#at 和loc用法大致相同,只是不支持切片!
print(s.at['a'])
#iat
print(s.iat[3])

#其他访问数据的方式
print("通过标签，类似数组索引访问元素的方法：",s['a'])

##类比ndarray中的布尔索引
print("通过布尔索引方式访问值小于二十的索引和值：",s[s<20])

##.head默认访问前五行数据
print(s.head())

##tail默认打印后五行
print(s.tail())
##上面两个都是如果括号里面不填数字就是默认打印五行，填了数字就按照数字要求的打印
```

### 三 常见函数

```python
#常见函数
import numpy as np
s=pd.Series([10,2,np.nan,None,3,4,5],index=['A','B','C','D','E','F','G'],name='data')
#查看所有的描述性信息
s.describe()
```

```python
#获取元素个数
print(s.count())
#获取索引
print(s.keys())  #通过方法获取（带小括号）
print(s.index) #通过属性获取

#判断哪个元素是缺失值
print(s.isna())

#检查每个元素是否在集合中
s.isin([4,5,6])
print(s)
```

```python
print(s.mean())#平均值
print(s.sum())#总和
print(s.std())#标准差
print(s.var())#方差
print(s.min())#最小值
print(s.max())#最大值
#Series都是.方法
```

```python
##排序
###按值排序
s1=s.sort_values()#**Jupyter 自动输出单元格最后一行表达式的返回值**，
#这个s并没有改变
print("s:\n",s)
print("s1:\n",s1)
###按索引排序
s2=s.sort_index()
print("按索引排序之后:\n",s2)
##分位数注意这个series里面的分位数是0-1的
print(s.quantile(0.75))

##众数
s['A']=2
print("众数：\n",s.mode())

##元素频率
# (每个元素出现多少次，并且会排好序，次数多的在前面)
print(s.value_counts())

##去重操作
s1=s.drop_duplicates()#这个操作并不会和改变原来的s
print(s1)

s2=s.unique()
print(s2)#这个去重方法得到的是ndarray，去掉了索引

##去重后元素的个数
print(s.nunique())
#count是统计除了缺失值之外的元素的个数
```
