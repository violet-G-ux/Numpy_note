## 1.绘制折线图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
##pyplot是 绘图库 matplotlib里面的子模块 专门封装好了简单的绘图函数

##字体 font
rcParams['font.sans-serif'] = ['SimHei']
#rcParams：全局运行配置参数
#**Runtime Configuration Parameters**，运行时配置参数

##创建图表，设置大小
plt.figure(figsize=(10,8))

##要绘图的数据
month=['1月','2月','3月','4月']
sales=[100,150,250,300]

##绘制折线图
plt.plot(month,sales,label='产品A',color='blue',linewidth=2,linestyle='--',marker='o')##这个label是图例的名字
##添加标题
plt.title('2025年销售趋势',color='red')
##添加坐标轴的标签
plt.xlabel('月份',fontsize=10)
plt.ylabel('销售额（万元）',fontsize=10)
##添加图例
plt.legend(loc='best')
##添加网格线
plt.grid(True,alpha=0.5,color='black')
##这个alpha是透明度的意思
##plt.grid(axis='x')
##设置刻度的字体大小
plt.xticks(rotation=0,fontsize=10)

##设置y轴的范围
plt.ylim(0,350)##y limit

##在每个数据点上显示数值
for x,y in zip(month,sales):
    plt.text(x,y+10,str(y),ha='center',va='bottom',fontsize=10)
##horizontal alignment 水平对齐方式
##vertical alignment  垂直对齐方式
##`plt.text(横坐标, 纵坐标, 文字内容)`：在图上写文字
##显示图表
plt.show()
```

## 2.绘制条形图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
##设置字体
rcParams['font.sans-serif'] = ['SimHei']
plt.figure(figsize=(10,8))
# 要绘图的数据
subjects = ['语文', '数学', '英语', '科学']
scores = [85, 92, 78, 88]
plt.ylabel('科目',fontsize=10)
plt.xlabel('成绩',fontsize=10)
##添加标题
plt.title('某人2025学年成绩分布',color='purple')
plt.grid(axis='x',alpha=0.1,color='black',linestyle='--')
##柱状图
#plt.bar(subjects,scores,label='小红',color='green')

##条形图
##注意是 barh
plt.barh(subjects,scores,label='小红')
plt.legend(loc='best')

##设置刻度的字体大小
plt.xticks(rotation=0,fontsize=10)

##设置x轴的范围
plt.xlim(0,100)##y limit

for x,y in zip(scores,subjects):
    plt.text(x+1,y,str(x),ha='left',va='bottom',fontsize=10)
##horizontal alignment 水平对齐方式
##vertical alignment  垂直对齐方式
##`plt.text(横坐标, 纵坐标, 文字内容)`：在图上写文字

##自动优化排版
plt.tight_layout()

##显示图表
plt.show()
```

## 5.绘制爆炸式饼图

## 6.绘制散点图

```python
import matplotlib.pyplot as plt
import random
from matplotlib import rcParams
##pyplot是 绘图库 matplotlib里面的子模块 专门封装好了简单的绘图函数

##字体 font
rcParams['font.sans-serif'] = ['SimHei']
#rcParams：全局运行配置参数
#**Runtime Configuration Parameters**，运行时配置参数

##创建图表，设置大小
plt.figure(figsize=(10,8))

##要绘图的数据

# scores=[50,55,60,65,70,75,80]
# hours=[1,2,3,4,5,6,7]
x=[]
y=[]
for i in range(1000):
    tmp=random.uniform(0,10)
    x.append(tmp)
    y.append(2.5*tmp+random.gauss(0,2))
plt.title('x y散点分布')
##绘制折线图
plt.scatter(x,y,label='散点',color='purple',alpha=0.4,s=5)

# ##添加坐标轴的标签
plt.xlabel('x变量',fontsize=10)
plt.ylabel('y变量',fontsize=10)
# ##图例位置
plt.legend(loc='best')
# ##添加网格线
plt.grid(True,alpha=0.5,color='black')
# ##这个alpha是透明度的意思
plt.grid(alpha=0.1,color='black',linestyle='--')
# ##设置刻度的字体大小
plt.xticks(rotation=0,fontsize=10)
#
# ##设置y轴的范围
plt.ylim(0,35)##y limit
#
# ##在每个数据点上显示数值
# for x,y in zip(month,sales):
#     plt.text(x,y+10,str(y),ha='center',va='bottom',fontsize=10)
# ##horizontal alignment 水平对齐方式
# ##vertical alignment  垂直对齐方式
# ##`plt.text(横坐标, 纵坐标, 文字内容)`：在图上写文字
#plt.tight_layout()
plt.plot([0,10],[0,25])
# ##显示图表
plt.show()
```

## 7.绘制箱线图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
##pyplot是 绘图库 matplotlib里面的子模块 专门封装好了简单的绘图函数

##字体 font
rcParams['font.sans-serif'] = ['SimHei']
# 模拟 3 门课的成绩
data = {
    '语文': [82, 85, 88, 70, 90, 76, 84, 83, 95],
    '数学': [75, 80, 79, 93, 88, 82, 87, 89, 92],
    '英语': [70, 72, 68, 65, 78, 80, 85, 90, 95]
}
plt.figure(figsize=(10,8))
plt.boxplot(data.values())
plt.xticks(ticks=range(1, len(data)+1), labels=data.keys(),fontsize=10)
plt.title('各科成绩分布')
plt.grid(True,alpha=0.5,color='black')
plt.ylabel('分数',fontsize=10)
plt.show()
```
