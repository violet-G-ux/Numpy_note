## 二.缺失值的判断和处理

```python
##缺失值的判断和处理
import pandas as pd
import numpy as np
s=pd.Series([1,2,np.nan,None,pd.NA])
df=pd.DataFrame([[1,pd.NA,2],[2,3,5],[None,4,6]],columns=['name','age','salary'])
print(s)

##查看是否是缺失值
print(s.isna())
print(s.isnull())
print(df.isna())
print(df.isnull())

##查看缺失值个数
print(s.isna().sum())
print(df.isna().sum())

```

```python
##剔除缺失值
print(s.dropna())
print(df.dropna())##只要一行中有一个缺失值就会把整行都剔除掉 这种默认 how='any'
print(df.dropna(how='all'))##只有一行都是缺失值的时候才剔除
print(df.dropna(thresh=2))##如果有n个及以上不是缺失值就保留
print(df.dropna(axis=1))##剔除一整列（注意！在drop方法里面 axis=1表示列 axis=0 表示行）与sum min max 等方法相反
print(df.dropna(subset=['age']))##指定检查哪些列，删行（这一行在指定列是空，删掉这一行）
```

```python
##填充缺失值
df=pd.read_csv('data/weather_withna.csv')
##使用字典的固定值填充
print(df.fillna({"max_temp":20}).head(10))

##使用统计值填充
print(df.fillna(df[['humidity']].mean()).head(10))

##根据缺失值附近的数值自动填充
print(df.ffill().tail()) #front
print(df.bfill().tail()) #behind
```

## 五 .数据变形

```python
import pandas as pd
data = {
    'ID': [1, 2],
    'name':['alice', 'bob'],
    'Math': [90, 85],
    'English': [88, 92],
    'Science': [95, 89]
}
df = pd.DataFrame(data)
df.T   #行列转置

'''
如果想要转换成如下形式
id  name    科目    成绩
1   alice   math    90
1   alice   English 88
1   alice   science 95
'''
##宽表转长表
df1=pd.melt(df,id_vars=['ID','name'],var_name='科目',value_name='分数')
#print(df1)
##再按照某列排序
df2=df1.sort_values('name')
print(df2)

##长标转宽表
df3=pd.pivot(df1,index=['ID','name'],columns='科目',values='分数')
print(df3)
```

## 六.数据分箱

```python
##等宽分箱（按数值区间）：pd.cut(要分箱的列, bins=边界列表, labels=标签列表)
##切数值，区间一样宽，人不一定一样多
import pandas as pd
df=pd.read_csv('data/employees.csv')
print(df.head())

### 选 1 列 → 一个中括号，Series
# 选 好几列 → 里面放列表，外层再加[]，两个中括号

pd.cut(df['salary'],bins=2) #bins =n  把这一列的数据分成两段
pd.cut(df['salary'],bins=2).value_counts()

##或者可以自己指定bins
pd.cut(df['salary'],bins=[1,10000,20000,30000,40000]).value_counts()
#加上lables
pd.cut(df['salary'],bins=[1,10000,20000,30000,40000],labels=['低薪','中薪','高薪','高高']).value_counts()
df['收入等级']=pd.cut(df['salary'],bins=[1,10000,20000,30000,40000],labels=['低薪','中薪','高薪','高高'])
print(df)
```

```python
### 等频分箱（按样本数量 / 分位数）：pd.qcut(要分箱的列, q=分组数量, labels=标签列表)
##切人数，每组人差不多，区间不一定一样宽
pd.qcut(df['salary'],q=2).value_counts()

'''
- 已经是文本分类：直接转 category，然后统计数量
- 是连续数字：先用 cut 分箱变成分类，再统计数量

##其实分箱最后就是把数值型转化成了类别型
'''
```

## 十.时间间隔

```python
# 时间间隔
d1 = pd.Timestamp('2013-01-15')
d2 = pd.Timestamp('2023-02-23')
d3 = d2 - d1
print(type(d3))
print(d3)


'''pd.date_range(start=None, end=None, periods=None, freq=None, tz=None, normalize=False, name=None, inclusive='both')

pd.date_range(...) 返回类型：DatetimeIndex

periods：一共要生成多少个时间点（数字）

freq：时间间隔（频率），隔多久生成一个

- `D` 天
- `h` 小时
- `min` 分钟
- `MS` 每月月初
- `ME` 每月月末
- `W` 每周

tz：时区，比如 'Asia/Shanghai' 北京时间

normalize：是否把时分秒清零。True 就统一变成 00:00:00

name：给生成出来的这组时间索引起个名字

inclusive：包不包含首尾边界
'both'：开头和结尾都包含（默认）
'left'：只包含开头，不要结尾
'right'：只包含结尾，不要开头


'''
# 示例：从2023-01-01开始，生成5个时间，间隔1天
dr = pd.date_range("2023-01-01", periods=5, freq="W")
print(dr)

```

## 十一.分组聚合

```python
# 分组聚合
# df.groupby('分组的字段')['聚合的字段'].聚合函数()
import pandas as pd
df = pd.read_csv('data/employees1.csv')
df=df.dropna(subset=['department_id'])
df['department_id']=df['department_id'].astype(int)
df.groupby('department_id').groups##查看分组
df.groupby('department_id').get_group(20)##查看某个分组里面的数据
df2=df.groupby('department_id')[['salary']].mean()##计算不同部门的平均薪资
##因为是两层中括号得到的是表所以可以取列
##如果只有一层中括号的话得到的是Series，只能通过索引进行取值
df2['salary']=df2['salary'].round(2)##保留两位小数
df2.reset_index()
df2.sort_values(by='salary', ascending=True, inplace=True)
df2
```

```python
##计算不同部门的不同岗位的人的平均薪资
df.groupby(['department_id','job_id']).groups##查看分组结果
df2=df.groupby(['department_id','job_id'])[['salary']].mean()

df2=df2.reset_index()
df2['salary']=df2['salary'].round(2)
print(df2)
```
